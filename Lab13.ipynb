```python
import pandas as pd
import plotly.express as px
```

```python
white_wine = pd.read_csv("winequality-white.csv", sep=";")
white_wine
```

```python
wine_scatter = px.scatter(white_wine,
                          x = "residual sugar",
                          y = "quality",
                          title = "Residual Sugar vs Quality in White Wine",
                          labels = {"quality":"Quality", "residual sugar":"Residual Sugar"}
                         )
wine_scatter
```
